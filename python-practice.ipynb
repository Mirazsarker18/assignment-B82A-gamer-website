```python
#Integers Integers, Floats, Booleans, Strings

x = 20
y = 100.10 
a = True
b = "Hassan Sarker"
c = False
print(type(x))
print(type(y))
print(type(a))
print(type(b))
print(type(c))
```

```python
# Floats exercise
flot_x = 20 / 9

print(flot_x)
print(type(flot_x))

# Round to 5 decimal places
flot_round = 40 / 7
round_flot = round(flot_round, 5)
print(round_flot)
print(type(round_flot))

```

```python
# Booleans
bool_exam = (2 < 40)
print(bool_exam)
print(type(bool_exam))

bool_exam2 = (40 < 5)
print(bool_exam2)
print(type(bool_exam2))

bool_not = not bool_exam2
print(bool_not)
print(type(bool_not))
```

```python
# Strings
str_exam = "125"
print(str_exam)
print(type(str_exam))

str_flot = "123.10"
print(str_flot)
print(type(str_flot))

```

```python
# convertible strings to flot
str_num = "1245.44"
print(type(str_num))
str_to_flot = float(str_num)
print(str_to_flot)
print(type(str_to_flot))
print("\n-------------")
str_sum = "Hassan " + "Sarker"
print(str_sum)
print(type(str_sum))
```
